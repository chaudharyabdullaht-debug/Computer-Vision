```python
import os
import shutil
import glob
import torchreid

# 1. Automatically locate where the dataset subfolders (bounding_box_train) exist
target_folders = ['bounding_box_train', 'bounding_box_test', 'query']
found_src_dir = None

# Search current directory and subdirectories for 'bounding_box_train'
for root_dir, dirs, files in os.walk('.'):
    if 'bounding_box_train' in dirs:
        found_src_dir = root_dir
        break

if not found_src_dir:
    raise FileNotFoundError(
        "Could not locate 'bounding_box_train'. Make sure the Kaggle dataset is fully unzipped."
    )

print(f"Found dataset files inside: {found_src_dir}")

# 2. Build the strict nested path torchreid expects: ./dukemtmc-reid/DukeMTMC-reID/
destination_dir = os.path.join('.', 'dukemtmc-reid', 'DukeMTMC-reID')
os.makedirs(destination_dir, exist_ok=True)

# 3. Move or copy the folders into the exact expected path
for folder in target_folders:
    src_path = os.path.join(found_src_dir, folder)
    dst_path = os.path.join(destination_dir, folder)

    if os.path.exists(src_path) and not os.path.exists(dst_path):
        print(f"Moving {folder} to {dst_path}...")
        shutil.move(src_path, dst_path)

# 4. Initialize ImageDataManager
datamanager = torchreid.data.ImageDataManager(
    root='.',                 # Current root directory containing 'dukemtmc-reid'
    sources='dukemtmcreid',   # Key name registered in torchreid
    targets='dukemtmcreid',
    height=256,
    width=128,
    batch_size_train=32,
    batch_size_test=100,
    transforms=['random_flip', 'random_crop']
)

print("\nDataset successfully loaded and verified!")
```

```python
import torch
import torchreid

# 1. Build OSNet model
model = torchreid.models.build_model(
    name='osnet_x1_0',
    num_classes=datamanager.num_train_pids,
    loss='softmax',
    pretrained=True
)

# Move model to GPU
model = model.cuda()

# 2. Setup Optimizer & Learning Rate Scheduler
optimizer = torchreid.optim.build_optimizer(
    model,
    optim='adam',
    lr=0.0003
)

scheduler = torchreid.optim.build_lr_scheduler(
    optimizer,
    lr_scheduler='single_step',
    stepsize=20
)

# 3. Create Engine for Training & Evaluation
engine = torchreid.engine.ImageSoftmaxEngine(
    datamanager,
    model,
    optimizer=optimizer,
    scheduler=scheduler,
    label_smooth=True
)
```

```python
# Run training and evaluation loop
engine.run(
    save_dir='log/osnet_dukemtmc',
    max_epoch=10,
    eval_freq=2,
    print_freq=20,
    test_only=False
)
```

```python
import numpy as np
import time

class MultiCameraTracker:
    def __init__(self, similarity_threshold=0.65, max_time_gap=15.0):
        self.similarity_threshold = similarity_threshold
        self.max_time_gap = max_time_gap  # Max seconds expected between camera transitions
        self.active_tracks = {}           # {global_id: {'feature': vector, 'last_cam': int, 'last_time': float}}
        self.next_global_id = 101
        self.blind_spots = []             # Logs identified blind spot transitions

    def update_detection(self, feature_vector, cam_id, current_time):
        matched_id = None
        highest_sim = -1.0

        # 1. Match against active global tracks using cosine similarity & temporal rules
        for g_id, track in self.active_tracks.items():
            time_diff = current_time - track['last_time']

            # Spatial-temporal constraint: ignore tracks that disappeared too long ago
            if time_diff > self.max_time_gap:
                continue

            # Calculate cosine similarity between normalized embeddings
            sim = np.dot(feature_vector, track['feature'])

            if sim > highest_sim and sim >= self.similarity_threshold:
                highest_sim = sim
                matched_id = g_id

        # 2. Case: Existing Person Re-Identified (No ID Change)
        if matched_id is not None:
            prev_cam = self.active_tracks[matched_id]['last_cam']
            time_spent = current_time - self.active_tracks[matched_id]['last_time']

            # Detect Blind Spot if person transitioned between non-overlapping cameras with a gap
            if prev_cam != cam_id and time_spent > 2.0:
                blind_spot_event = {
                    'global_id': matched_id,
                    'from_camera': prev_cam,
                    'to_camera': cam_id,
                    'time_in_blind_spot': round(time_spent, 2)
                }
                self.blind_spots.append(blind_spot_event)
                print(f"[BLIND SPOT DETECTED] Person {matched_id} was unmonitored for {time_spent:.1f}s between Camera {prev_cam} -> Camera {cam_id}")

            # Update person's latest feature and camera location
            self.active_tracks[matched_id] = {
                'feature': feature_vector,
                'last_cam': cam_id,
                'last_time': current_time
            }
            print(f"[MATCH] Person identified as Global ID_{matched_id} at Camera {cam_id} (Sim: {highest_sim:.4f})")
            return matched_id

        # 3. Case: New Person Enters Store
        else:
            assigned_id = self.next_global_id
            self.next_global_id += 1
            self.active_tracks[assigned_id] = {
                'feature': feature_vector,
                'last_cam': cam_id,
                'last_time': current_time
            }
            print(f"[NEW ENTRY] Assigned Global ID_{assigned_id} at Camera {cam_id}")
            return assigned_id

# Instantiate tracker
tracker = MultiCameraTracker(similarity_threshold=0.60, max_time_gap=20.0)

# Run a test simulation across 3 camera detections
t0 = time.time()
id1 = tracker.update_detection(feat1, cam_id=1, current_time=t0)
id2 = tracker.update_detection(feat2, cam_id=2, current_time=t0 + 5.5)  # 5.5s delay -> triggers blind spot check
```

```python
import cv2
import torch
import numpy as np
from ultralytics import YOLO  # For Stage 1: Detection
import time

# ----------------------------------------------------
# 1. SETUP MODELS (Detector + OSNet Re-ID)
# ----------------------------------------------------
# Stage 1: Detector
detector = YOLO('yolov8n.pt')  # Pre-trained YOLO person detector

# Stage 3: OSNet Re-ID Feature Extractor (Loaded from earlier steps)
model.eval()
test_transform = datamanager.transform_te

# ----------------------------------------------------
# 2. FEATURE EXTRACTION FUNCTION
# ----------------------------------------------------
def get_crop_embedding(crop_img):
    """Extracts a 512-d feature vector from a cropped person image."""
    from PIL import Image
    pil_img = Image.fromarray(cv2.cvtColor(crop_img, cv2.COLOR_BGR2RGB))
    img_tensor = test_transform(pil_img).unsqueeze(0).cuda()

    with torch.no_grad():
        features = model(img_tensor)
        features = torch.nn.functional.normalize(features, p=2, dim=1)
    return features.cpu().numpy().flatten()

# ----------------------------------------------------
# 3. VIDEO PROCESSING & LOGGING LOOP
# ----------------------------------------------------
def process_store_video(video_path, camera_id):
    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
    frame_idx = 0

    # Dictionary to store timestamps: {global_id: {'entry': time, 'exit': time}}
    person_logs = {}

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        frame_idx += 1
        current_time_sec = frame_idx / fps

        # --- Stage 1: Run YOLO Person Detection ---
        results = detector(frame, classes=[0], verbose=False)[0]  # class 0 = person

        for box in results.boxes.xyxy.cpu().numpy():
            x1, y1, x2, y2 = map(int, box)

            # Crop detected person from frame
            person_crop = frame[y1:y2, x1:x2]
            if person_crop.size == 0:
                continue

            # --- Stage 2 & 3: Extract Embedding & Assign Global ID ---
            embedding = get_crop_embedding(person_crop)

            # Match embedding against Global Tracker
            global_id = tracker.update_detection(
                feature_vector=embedding,
                cam_id=camera_id,
                current_time=current_time_sec
            )

            # --- Log Entry and Exit Times ---
            if global_id not in person_logs:
                person_logs[global_id] = {
                    'entry_time': round(current_time_sec, 2),
                    'exit_time': round(current_time_sec, 2)
                }
                print(f"⏱️ [LOG] Person ID_{global_id} ENTERED at {current_time_sec:.2f}s (Cam {camera_id})")
            else:
                # Update exit time dynamically as long as they are visible
                person_logs[global_id]['exit_time'] = round(current_time_sec, 2)

            # Draw Bounding Box & Unique ID on Video Output
            cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(frame, f"ID: {global_id}", (x1, y1 - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    cap.release()
    return person_logs

# ----------------------------------------------------
# 4. EXECUTE PIPELINE
# ----------------------------------------------------
# Run pipeline on a sample video
logs = process_store_video('store_camera_feed.mp4', camera_id=1)

# Summary of tracking logs
print("\n=== FINAL ENTRY/EXIT LOGS ===")
for pid, info in logs.items():
    duration = info['exit_time'] - info['entry_time']
    print(f"Person ID_{pid} | Entry: {info['entry_time']}s | Exit: {info['exit_time']}s | Duration: {duration:.2f}s")
```

```python
import cv2
import numpy as np
from ultralytics import YOLO

# 1. Load YOLOv8 model
model = YOLO('yolov8n.pt')

# 2. Open input video
cap = cv2.VideoCapture(video_path)

fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# 3. Configure Output VideoWriter
output_path = 'tracked_blindspot_output.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

# 4. Define Blind Spot Region (Polygon Coordinates: [x, y])
# Customize these coordinates according to your video frame dimensions
blind_spot_polygon = np.array([
    [0, 0],
    [int(width * 0.35), 0],
    [int(width * 0.2), int(height * 0.4)],
    [0, int(height * 0.5)]
], np.int32)

print("Processing video tracking with blind spots...")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # --- TRACKING ---
    # classes=[0] filters ONLY persons
    # persist=True maintains unique tracking IDs across frames
    # tracker="bytetrack.yaml" handles persistent identification
    results = model.track(frame, persist=True, classes=[0], tracker="bytetrack.yaml", verbose=False)

    annotated_frame = frame.copy()

    # Draw person detections with persistent IDs
    if results[0].boxes is not None and results[0].boxes.id is not None:
        boxes = results[0].boxes.xyxy.cpu().numpy()
        track_ids = results[0].boxes.id.cpu().numpy().astype(int)

        for box, track_id in zip(boxes, track_ids):
            x1, y1, x2, y2 = map(int, box)

            # Draw green bounding box
            cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Label with Unique Person ID
            label = f"ID: {track_id}"
            cv2.putText(annotated_frame, label, (x1, max(20, y1 - 10)),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    # --- BLIND SPOT OVERLAY ---
    # Create semi-transparent red overlay zone
    overlay = annotated_frame.copy()
    cv2.fillPoly(overlay, [blind_spot_polygon], (0, 0, 255))  # Red fill

    # Blend overlay with frame (0.4 transparency)
    cv2.addWeighted(overlay, 0.4, annotated_frame, 0.6, 0, annotated_frame)

    # Outline the blind spot border
    cv2.polylines(annotated_frame, [blind_spot_polygon], isClosed=True, color=(0, 0, 255), thickness=2)
    cv2.putText(annotated_frame, "BLIND SPOT", (blind_spot_polygon[0][0] + 10, blind_spot_polygon[0][1] + 30),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

    out.write(annotated_frame)

cap.release()
out.release()
print(f"Finished! File saved to {output_path}")
```

```python
import os
import cv2
import torch
import numpy as np
from PIL import Image
from ultralytics import YOLO
from google.colab import files

# 1. Initialize YOLO detector separately
yolo_detector = YOLO('yolov8n.pt')

# 2. Persistent Re-ID Tracker
class PersistentReIDTracker:
    def __init__(self, reid_model, transform, similarity_threshold=0.55, max_time_gap=30.0):
        self.reid_model = reid_model
        self.transform = transform
        self.similarity_threshold = similarity_threshold
        self.max_time_gap = max_time_gap
        self.gallery = {}
        self.next_global_id = 1

    def match_or_create_id(self, crop_img, current_time):
        # Convert BGR crop to PIL RGB Image
        pil_img = Image.fromarray(cv2.cvtColor(crop_img, cv2.COLOR_BGR2RGB))

        # Apply Torchreid test transforms (Normalizes to 0.0-1.0 and applies ImageNet mean/std)
        img_tensor = self.transform(pil_img).unsqueeze(0).cuda()

        with torch.no_grad():
            # Pass crop tensor through OSNet feature extractor
            features = self.reid_model(img_tensor)

            # Handle list/tuple output structure from Torchreid
            if isinstance(features, (list, tuple)):
                features = features[0]

            # Flatten feature map to 1D vector (e.g., [1, 512] -> [512])
            if features.dim() > 2:
                features = features.view(features.size(0), -1)

            # L2 Normalize embedding vector
            feature_tensor = torch.nn.functional.normalize(features, p=2, dim=1)
            feature_vector = feature_tensor.cpu().numpy().flatten()

        matched_id = None
        highest_sim = -1.0

        # Match embedding against existing active gallery
        for g_id, record in self.gallery.items():
            time_diff = current_time - record['last_time']
            if time_diff > self.max_time_gap:
                continue

            sim = np.dot(feature_vector, record['feature'])
            if sim > highest_sim and sim >= self.similarity_threshold:
                highest_sim = sim
                matched_id = g_id

        # Update track feature vector using Moving Average or register new ID
        if matched_id is not None:
            old_feat = self.gallery[matched_id]['feature']
            updated_feat = 0.7 * old_feat + 0.3 * feature_vector
            updated_feat = updated_feat / np.linalg.norm(updated_feat)

            self.gallery[matched_id]['feature'] = updated_feat
            self.gallery[matched_id]['last_time'] = current_time
            return matched_id
        else:
            assigned_id = self.next_global_id
            self.next_global_id += 1
            self.gallery[assigned_id] = {
                'feature': feature_vector,
                'last_time': current_time
            }
            return assigned_id

# Instantiate tracker passing explicit OSNet instance & transform
persistent_tracker = PersistentReIDTracker(
    reid_model=osnet_model,
    transform=datamanager.transform_te,
    similarity_threshold=0.55,
    max_time_gap=30.0
)

# 3. Video Pipeline Execution
cap = cv2.VideoCapture(video_path)

fps = int(cap.get(cv2.CAP_PROP_FPS)) or 30
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

temp_output = 'temp_raw_output.mp4'
whatsapp_output = 'whatsapp_tracked_reid.mp4'

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(temp_output, fourcc, fps, (width, height))

# Polygon overlay for blind spot region
blind_spot_polygon = np.array([
    [0, 0],
    [int(width * 0.35), 0],
    [int(width * 0.2), int(height * 0.4)],
    [0, int(height * 0.5)]
], np.int32)

frame_idx = 0
print("Running persistent tracking pipeline...")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    frame_idx += 1
    current_time_sec = frame_idx / fps

    # Run YOLO detection only on person class (class 0)
    yolo_results = yolo_detector(frame, classes=[0], conf=0.35, verbose=False)[0]
    annotated_frame = frame.copy()

    for box in yolo_results.boxes.xyxy.cpu().numpy():
        x1, y1, x2, y2 = map(int, box)

        # Ignore invalid small bounding box crops
        person_crop = frame[y1:y2, x1:x2]
        if person_crop.size == 0 or person_crop.shape[0] < 15 or person_crop.shape[1] < 15:
            continue

        # Extract embedding and match/assign global persistent ID
        persistent_id = persistent_tracker.match_or_create_id(person_crop, current_time_sec)

        # Draw green bounding boxes and global tracking label
        cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(annotated_frame, f"Person ID: {persistent_id}", (x1, max(20, y1 - 10)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    # Render Blind Spot Polygon Region
    overlay = annotated_frame.copy()
    cv2.fillPoly(overlay, [blind_spot_polygon], (0, 0, 255))
    cv2.addWeighted(overlay, 0.35, annotated_frame, 0.65, 0, annotated_frame)
    cv2.polylines(annotated_frame, [blind_spot_polygon], isClosed=True, color=(0, 0, 255), thickness=2)
    cv2.putText(annotated_frame, "BLIND SPOT", (blind_spot_polygon[0][0] + 10, blind_spot_polygon[0][1] + 25),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)

    out.write(annotated_frame)

cap.release()
out.release()

print("Encoding video for WhatsApp (H.264 / yuv420p)...")
os.system(f"ffmpeg -y -i {temp_output} -vcodec libx264 -pix_fmt yuv420p -acodec aac -movflags +faststart {whatsapp_output}")

print(f"Done! Downloading {whatsapp_output}...")
files.download(whatsapp_output)
```

```python
class GalleryQueueTracker:
    def __init__(self, reid_model, transform, similarity_threshold=0.52, max_gallery_size=5):
        self.reid_model = reid_model
        self.transform = transform
        self.similarity_threshold = similarity_threshold
        self.max_gallery_size = max_gallery_size
        self.gallery = {}  # {global_id: {'features': [feat1, feat2, ...], 'last_time': t}}
        self.next_global_id = 1

    def match_or_create_id(self, crop_img, current_time):
        feature_vector = self._extract_feature(crop_img)

        matched_id = None
        highest_sim = -1.0

        for g_id, record in self.gallery.items():
            if current_time - record['last_time'] > 30.0:
                continue

            # Compare new feature against ALL stored historical views of this person
            sims = [np.dot(feature_vector, f) for f in record['features']]
            max_sim_for_id = max(sims)

            if max_sim_for_id > highest_sim and max_sim_for_id >= self.similarity_threshold:
                highest_sim = max_sim_for_id
                matched_id = g_id

        if matched_id is not None:
            # Update last seen time
            self.gallery[matched_id]['last_time'] = current_time

            # If the new view is distinct enough (e.g., sim < 0.85), add it as a new pose variant
            if highest_sim < 0.85:
                self.gallery[matched_id]['features'].append(feature_vector)
                if len(self.gallery[matched_id]['features']) > self.max_gallery_size:
                    self.gallery[matched_id]['features'].pop(0)  # Evict oldest vector
            return matched_id
        else:
            assigned_id = self.next_global_id
            self.next_global_id += 1
            self.gallery[assigned_id] = {
                'features': [feature_vector],
                'last_time': current_time
            }
            return assigned_id
```

```python
def calculate_spatial_similarity(box1, box2, frame_w, frame_h):
    # Box centroids
    c1 = np.array([(box1[0] + box1[2]) / 2, (box1[1] + box1[3]) / 2])
    c2 = np.array([(box2[0] + box2[2]) / 2, (box2[1] + box2[3]) / 2])

    # Euclidean distance normalized by frame diagonal
    dist = np.linalg.norm(c1 - c2)
    max_dist = np.sqrt(frame_w**2 + frame_h**2)

    # Convert distance to similarity (1 = close, 0 = far)
    return max(0.0, 1.0 - (dist / (max_dist * 0.25)))
```

```python
 import os
import cv2
import torch
import numpy as np
from PIL import Image
from ultralytics import YOLO
from google.colab import files

# 1. LOAD YOLO MODEL FOR BYTE TRACK / BOT-SORT TRACKING
# Ultralytics natively supports BoT-SORT (botsort.yaml) and ByteTrack (bytetrack.yaml)
yolo_detector = YOLO('yolov8n.pt')

# 2. OPTIONAL: OSNET GLOBAL RE-ID RE-ASSOCIATION ENGINE
# Built-in trackers handle frame-to-frame tracking and brief occlusions exceptionally well.
# We use OSNet as a global persistent Re-ID fall-back when local tracks break.
class HybridReIDTracker:
    def __init__(self, reid_model, transform, similarity_threshold=0.58, max_time_gap=45.0):
        self.reid_model = reid_model
        self.transform = transform
        self.similarity_threshold = similarity_threshold
        self.max_time_gap = max_time_gap
        self.local_to_global_map = {}  # Maps YOLO track_id -> Persistent Global ID
        self.gallery = {}              # {global_id: {'features': [f1, f2], 'last_time': t}}
        self.next_global_id = 1

    def get_persistent_id(self, crop_img, track_id, current_time):
        # Return mapped persistent global ID if local tracker track_id is already assigned
        if track_id in self.local_to_global_map:
            gid = self.local_to_global_map[track_id]
            self.gallery[gid]['last_time'] = current_time
            return gid

        # Extract OSNet feature embedding for unmapped or recovered tracks
        pil_img = Image.fromarray(cv2.cvtColor(crop_img, cv2.COLOR_BGR2RGB))
        img_tensor = self.transform(pil_img).unsqueeze(0).cuda()

        with torch.no_grad():
            features = self.reid_model(img_tensor)
            if isinstance(features, (list, tuple)):
                features = features[0]
            if features.dim() > 2:
                features = features.view(features.size(0), -1)

            feature_tensor = torch.nn.functional.normalize(features, p=2, dim=1)
            feature_vector = feature_tensor.cpu().numpy().flatten()

        matched_gid = None
        highest_sim = -1.0

        # Match against global gallery multi-vector views
        for g_id, record in self.gallery.items():
            if current_time - record['last_time'] > self.max_time_gap:
                continue

            sims = [np.dot(feature_vector, f) for f in record['features']]
            max_sim = max(sims)

            if max_sim > highest_sim and max_sim >= self.similarity_threshold:
                highest_sim = max_sim
                matched_gid = g_id

        # Map to matched global ID or create new global identity entry
        if matched_gid is not None:
            self.local_to_global_map[track_id] = matched_gid
            self.gallery[matched_gid]['last_time'] = current_time

            if highest_sim < 0.85 and len(self.gallery[matched_gid]['features']) < 5:
                self.gallery[matched_gid]['features'].append(feature_vector)
            return matched_gid
        else:
            assigned_gid = self.next_global_id
            self.next_global_id += 1
            self.local_to_global_map[track_id] = assigned_gid
            self.gallery[assigned_gid] = {
                'features': [feature_vector],
                'last_time': current_time
            }
            return assigned_gid


# Instantiate Persistent Hybrid Re-ID Engine
persistent_reid = HybridReIDTracker(
    reid_model=osnet_model,
    transform=datamanager.transform_te,
    similarity_threshold=0.58,
    max_time_gap=45.0
)

# 3. VIDEO PROCESSING PIPELINE
cap = cv2.VideoCapture(video_path)

fps = int(cap.get(cv2.CAP_PROP_FPS)) or 30
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

temp_output = 'temp_raw_output.mp4'
whatsapp_output = 'whatsapp_tracked_botsort.mp4'

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(temp_output, fourcc, fps, (width, height))

# Polygon overlay region for blind spot
blind_spot_polygon = np.array([
    [0, 0],
    [int(width * 0.35), 0],
    [int(width * 0.2), int(height * 0.4)],
    [0, int(height * 0.5)]
], np.int32)

frame_idx = 0
print("Running BoT-SORT / DeepOCSORT with OSNet Re-ID integration...")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    frame_idx += 1
    current_time_sec = frame_idx / fps

    # Execute Ultralytics BoT-SORT tracker (use "bytetrack.yaml" for ByteTrack)
    results = yolo_detector.track(
        source=frame,
        persist=True,
        tracker="botsort.yaml",
        classes=[0],
        conf=0.35,
        verbose=False
    )[0]

    annotated_frame = frame.copy()

    # Process detections if active tracked boxes exist
    if results.boxes is not None and results.boxes.id is not None:
        boxes = results.boxes.xyxy.cpu().numpy()
        track_ids = results.boxes.id.int().cpu().numpy()

        for box, track_id in zip(boxes, track_ids):
            x1, y1, x2, y2 = map(int, box)

            person_crop = frame[y1:y2, x1:x2]
            if person_crop.size == 0:
                continue

            h, w = person_crop.shape[:2]
            aspect_ratio = h / float(w) if w > 0 else 0

            # Filter tiny crops or unnatural bounding box proportions
            if h < 35 or w < 18 or aspect_ratio < 1.1:
                continue

            # Resolve global persistent ID across long-term occlusions/blind spots
            global_id = persistent_reid.get_persistent_id(person_crop, track_id, current_time_sec)

            # Render bounding box and persistent tracking tag
            cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(
                annotated_frame,
                f"ID: {global_id} (Trk: {track_id})",
                (x1, max(20, y1 - 10)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2
            )

    # Render Blind Spot Polygon Region
    overlay = annotated_frame.copy()
    cv2.fillPoly(overlay, [blind_spot_polygon], (0, 0, 255))
    cv2.addWeighted(overlay, 0.35, annotated_frame, 0.65, 0, annotated_frame)
    cv2.polylines(annotated_frame, [blind_spot_polygon], isClosed=True, color=(0, 0, 255), thickness=2)
    cv2.putText(annotated_frame, "BLIND SPOT", (blind_spot_polygon[0][0] + 10, blind_spot_polygon[0][1] + 25),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)

    out.write(annotated_frame)

cap.release()
out.release()

print("Encoding video for WhatsApp (H.264 / yuv420p)...")
os.system(f"ffmpeg -y -i {temp_output} -vcodec libx264 -pix_fmt yuv420p -acodec aac -movflags +faststart {whatsapp_output}")

print(f"Done! Downloading {whatsapp_output}...")
files.download(whatsapp_output)
```
